```python
import pandas
```

```python
dataset = pandas.read_csv('wine.csv')
```

```python
dataset.info()
```

```python
dataset
```

```python
y = dataset['Wine']
```

```python
y
```

```python
dataset.columns
```

```python
y_cat = pandas.get_dummies(y)
```

```python
# Neural network selects the feature from X automatically. No need to do feature elemination manually.
```

```python
X = dataset[  ['Alcohol', 'Malic.acid', 'Ash', 'Acl', 'Mg', 'Phenols',
       'Flavanoids', 'Nonflavanoid.phenols', 'Proanth', 'Color.int', 'Hue',
       'OD', 'Proline']  ]
```

```python
y_cat
```

```python
X
```

```python
from keras.models import Sequential
```

```python
from keras.layers import Dense
```

```python
model = Sequential()
```

```python
model.add(   
        Dense(
                units=10, 
                input_dim=13,
                activation="relu",
                bias_initializer="zeros",
                kernel_initializer="zeros"
        
                )
        )
```

```python
model.summary()
```

```python
model.add(   
        Dense(
                units=8, 
                activation="relu",
                bias_initializer="zeros",
                kernel_initializer="zeros"
        
                )
        )
```

```python
model.summary()
```

```python
model.layers[1]
```

```python
model.layers[1].get_weights()
```

```python
model.layers[0].output
```

```python
model.layers[0].input
```

```python
model.add(   
        Dense(
                units=8, 
                activation="relu",
                bias_initializer="zeros",
                kernel_initializer="zeros"
        
                )
        )
```

```python
model.summary()
```

```python
model.add(   
        Dense(
                units=3, 
                activation="softmax",
                bias_initializer="zeros",
                kernel_initializer="zeros"
        
                )
        )
```

```python
model.summary()
```

```python
model.compile(loss="categorical_crossentropy")
```

```python
model.fit(X, y_cat)
```

```python
model.get_weights()
```

```python
model.predict(X)
```
