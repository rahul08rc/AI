```python
import pandas
```

```python
dataset1 = pandas.read_csv('500_Person_Gender_Height_Weight_Index.csv')
```

```python
dataset1.columns
```

```python
dataset = dataset1[[  'Gender', 'Height', 'Weight'   ]]
```

```python
dataset
```

```python
dataset.describe()
```

```python
# Here Height will decide Weight
# y = Weight
```

```python
y = dataset[  'Weight'  ]
```

```python
X = dataset[  'Height'  ]
```

```python
from keras.models import Sequential
```

```python
from keras.layers import Dense
```

```python
# brain = Artificial Nural Network(ANN)

brain = Sequential()
```

```python
brain.get_config()
```

```python
# Making Layer
# By default, activation will be LINEAR

brain.add( Dense( units=1, activation="linear" ) )
```

```python
brain.get_config()
```

```python
brain.get_layer
```

```python
brain.get_weights()
```

```python
# brain.fit(X , y)
```

```python
brain.compile(loss='mean_absolute_error'  )
```

```python
brain.fit(X , y, epochs=6)
```

```python
brain.predict
```

```python
brain.get_weights()
```
