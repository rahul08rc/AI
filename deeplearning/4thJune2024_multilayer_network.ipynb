```python
import pandas
```

```python
dataset = pandas.read_csv('Churn_Modelling.csv')
```

```python
dataset
```

```python
dataset.columns
```

```python
y = dataset[  'Exited' ]
```

```python
# X = dataset[ [  'CreditScore', 'Geography',
  #     'Gender', 'Age', 'Tenure', 'Balance', 'NumOfProducts', 'HasCrCard',
   #    'IsActiveMember', 'EstimatedSalary'  ]   ]
```

```python


geo = pandas.get_dummies(  dataset['Geography'], drop_first=True  )
```

```python
gender = pandas.get_dummies(  dataset['Gender'], drop_first=True  )
```

```python
tenure = pandas.get_dummies(  dataset['Tenure'], drop_first=True  )
```

```python
product = pandas.get_dummies(  dataset['NumOfProducts'], drop_first=True  )
```

```python
crCard = pandas.get_dummies(  dataset['HasCrCard'], drop_first=True  )
```

```python
activeMenber = pandas.get_dummies(  dataset['IsActiveMember'], drop_first=True  )
```

```python
X = pandas.concat(  [  dataset['CreditScore'], dataset['Age'], dataset['Balance'], dataset['EstimatedSalary'], geo, gender, tenure, product, crCard, activeMenber ], axis=1  )
```

```python
X
```

```python
# Last layer gives the outcome
```

```python
from keras.models import Sequential
```

```python
brain = Sequential()
```

```python
brain.get_config
```

```python
brain.summary()
```

```python
from keras.layers import Dense
```

```python
# Input Dimension is 22. Because 22 columns are there. input_dim=22.   Need to mention in the first layer.
# Here we are adding 5 Neurons. Hence units=5
# It will be the first Layer.

brain.add(  
    Dense(  
            input_dim=22, 
            units=5, 
            activation='relu' ,
            kernel_initializer="zeros" ,
            bias_initializer="zeros"
    )
)
```

```python
# Param = (22 + 1) * 5
# For eacn unit(nuron), 22 coefficient and 1 bias.

brain.summary()
```

```python
# Layer 2
# 5 input from 1st layer
# Param = (5 + 1) * 5 = 30

brain.add(   Dense(units=5, activation='relu', kernel_initializer="zeros", bias_initializer="zeros") )
```

```python
brain.summary()
```

```python
# Layer 3

brain.add(   Dense(units=3  , activation='relu', kernel_initializer="zeros", bias_initializer="zeros") )
```

```python
brain.summary()
```

```python
# Layer 4

brain.add(   Dense(units=1  , activation='sigmoid') )
```

```python
brain.summary()
```

```python
brain.get_config()
```

```python
# Binary classification lass function Keras

brain.compile(loss="binary_crossentropy")
```

```python
brain.fit(X, y, epochs=20)
```
