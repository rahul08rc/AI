```python
dataset = pandas.read_csv('50_Startups.csv')
```

```python
y = dataset['Profit']
```

```python
dataset.columns
```

```python
x = dataset[ ['R&D Spend', 'Administration', 'Marketing Spend'] ]
```

```python
state = dataset['State']
```

```python
# In Dummy variables, we always remove a feature(column) dependent on another column. Categorical variable. 
# We need to add "drop_first=True" for the dependent Feature Elimination.

dummy_state = pandas.get_dummies(state, drop_first=True)
```

```python
# This is a Dummy Variable & This type is Boolean. True = 1  & False = 0
# dummy_state is also a part of x

dummy_state
```

```python
# We already have value for "x". And we added Florida & New York inside x. Means, take dataset from "dummy_state"


x[ ["Florida", "New York"]  ] = dummy_state
```

```python
x
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x, y)
```

```python
model.coef_
```
