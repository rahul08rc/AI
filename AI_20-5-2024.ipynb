```python
import pandas
```

```python
dataset1 = pandas.read_csv('SalaryData.csv')
```

```python
# Here values are coming continuously. It is called Linear Regression

dataset1
```

```python
dataset1.info()
```

```python
y = dataset1['Salary']
```

```python
x = dataset1['YearsExperiance']
```

```python
type(x)
```

```python
# x.values convert pandas to numpy. reshape(-1 , 1)= It converts 1D data to 2D
X = x.values.reshape(-1, 1)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model1 = LinearRegression()
```

```python
# Till now model is not trained.
```

```python
# Hit and Trial Method

model1.fit(X , y)
```

```python
model1.coef_
```

```python
# This model is not accurate. As per this model, Fresher will get zero(0) salary
#But for freshers, this model has to be BIASed.
# here y = mx is not applicable. We have to use y = ms + b(b = bias)
```

```python
model1.predict( [[ 1 ]] )
```
