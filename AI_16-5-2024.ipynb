```python
import pandas
```

```python
# marks.csv is loaded in jupyter

dataset = pandas.read_csv('marks.csv')
```

```python
#  marks = y = output

dataset
```

```python
X = dataset['hrs']
```

```python
X
```

```python
y = dataset['marks']
```

```python
y
```

```python
type(X)
```

```python
x = X.values
```

```python
x
```

```python
# Making the array 2D

p = x.reshape(-1, 1)
```

```python
p
```

```python
from sklearn.linear_model import LinearRegression
```

```python
mind = LinearRegression()
```

```python
# Function:: we need to write y at the end since finding the mark is our end goal
# model trained: by learning from X and y

mind.fit(p , y)
```

```python


mind.predict( [[9]]  )
```

```python
mind.coef_
```

```python
 # to export code

import joblib
```

```python
joblib.dump(mind, "marks_model")
```
